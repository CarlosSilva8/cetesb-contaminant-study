```python
# Manipulação de dados
import pandas as pd

# Visualização de dataframe
from IPython.display import display

# Confecção de plots
import matplotlib.pyplot as plt
import seaborn as sns

# Manipulação de dados geográficos
import geopandas as gpd
```

```python
# Função de visualização de dataframes

def df_visualization(df):
    print(f'Formato: {df.shape}\n\nInformações:')
    print(df.info())
    display(df.head(2))


# Função de fazer uma soma de dados agregados de um dataframe
def sum_aggregate_function(df, indice1, indice2, col_somada):
    if indice2 != 0:
        x = df.groupby([indice1] + indice2)[col_somada].sum()
    elif indice2 == 0:
        x = df.groupby(indice1)[col_somada].sum()
    
    return x
```

```python
# Criação do dataframe 'cru'
excel_path = 'C:/Users/User/Documents/Códigos Github/cetesb-contaminant-study/notebooks/areas_contaminadas_modelagem.xlsx'
df = pd.read_excel(excel_path)
df = pd.DataFrame(df)

# Visualização inicial do dataframe
df_visualization(df)
```

```python
# Retirada de colunas indesejadas
cols_to_drop = ['OBJECTID',	'NIS', 'IdMunSipol']
df = df.drop(columns=cols_to_drop)
df_visualization(df)
```

```python
# Leitura e visualização do shapefile de SP
shape_path = "C:/Users/User/Documents/Códigos Github/cetesb-contaminant-study/SP_Municipios_2024/SP_Municipios_2024.shp"
geo_df = gpd.read_file(shape_path)
df_visualization(geo_df)
```

```python
# Certificando que o nome e tipo de dado de CD_MUN seja o mesmo do que CodIBGE
geo_df = geo_df.rename(columns={'CD_MUN': 'CodIBGE'})
geo_df['CodIBGE'] = geo_df['CodIBGE'].astype(int)
geo_df = geo_df[['CodIBGE', 'geometry']]

# Merge (equivalente a um left join) para inserir dados georreferenciados no dataframe
geo_df = df.merge(geo_df, how='left', left_on='CodIBGE', right_on='CodIBGE')


df_visualization(geo_df)
```

```python
# Listando as colunas dummies para facilitar agrupamentos
cols_contaminantes = [col for col in geo_df.columns if col.startswith('grupo_contaminante_')]
cols_atividades = [col for col in geo_df.columns if col.startswith('grupo_atividade_')]
cols_fontes = [col for col in geo_df.columns if col.startswith('fonte_contaminacao')]
cols_meios = [col for col in geo_df.columns if col.startswith('meio_impactado')]

# Visualização da quantidade de elementos em cada lista
print(f'Grupos de contaminantes: {len(cols_contaminantes)}\n'
      f'Grupos de atividades: {len(cols_atividades)}\n'
      f'Fontes de contaminação: {len(cols_fontes)}\n'
      f'Meios impactados: {len(cols_meios)}\n')

```

```python
# Agregação das informações por município
    # Contaminantes
contaminantes_per_mun = sum_aggregate_function(geo_df, 'NomMunicipio', 0, cols_contaminantes)
    # Fontes de Contaminação
fontes_per_mun = sum_aggregate_function(geo_df, 'NomMunicipio', 0, cols_fontes)
    # Meios Impactados
meio_per_mun = sum_aggregate_function(geo_df, 'NomMunicipio', 0, cols_meios)
    # Grupos de Atividade
atividade_per_mun = sum_aggregate_function(geo_df, 'NomMunicipio', 0, cols_atividades)



# Agregação das informações por município e atividade
    # Contaminantes
contaminantes_per_atv = sum_aggregate_function(geo_df, 'NomMunicipio', cols_atividades, cols_contaminantes)
    # Fontes de Contaminação
fontes_per_atv = sum_aggregate_function(geo_df, 'NomMunicipio', cols_atividades, cols_fontes)
    # Meios Impactados
meio_per_atv = sum_aggregate_function(geo_df, 'NomMunicipio', cols_atividades, cols_meios)

# Visualização da agregação por município
display(contaminantes_per_mun.sort_values(by=cols_contaminantes).tail(1))
display(fontes_per_mun.sort_values(by=cols_fontes).tail(1))
display(meio_per_mun.sort_values(by=cols_meios).tail(1))
display(atividade_per_mun.sort_values(by=cols_atividades).tail(1))

print(
    f'\n----------------------------------------------------------------'
    '--------------------------------------------------------------------------------------'
    f'\nSegunda leva de dataframes:\n'
    f'------------------------------------------------------------------'
    '--------------------------------------------------------------------------------------\n')

# Visualização da agregação por município E atividade
display(contaminantes_per_atv.sort_values(by=cols_contaminantes).tail(1))
display(fontes_per_atv.sort_values(by=cols_fontes).tail(1))
display(meio_per_atv.sort_values(by=cols_meios).tail(1))
```
